```python
Q1 = df['shares'].quantile(0.25)
Q3 = df['shares'].quantile(0.75)
IQR = Q3 - Q1

Q1, Q3, IQR
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

lower_bound, upper_bound
```

```python
df_no_outliers = df[
    (df['shares'] >= lower_bound) & 
    (df['shares'] <= upper_bound)
]
```

```python
df_no_outliers['shares'].mean()
```
